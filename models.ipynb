```python
import numpy as np
import pandas as pd
```

```python
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
```

```python
from clean import DataCleaner
import preprocess
```

```python
df = pd.read_csv('data/bike_hour.csv')
```

```python
cleaner = DataCleaner(df)
df = cleaner.clean()
```

```python
df = preprocess.convert_date(df)
df = preprocess.convert_weather_units(df)
df = preprocess.add_cyclical(df, 'mnth', 12)
```

```python
features = ['hr', 'mnth', 'mnth_sin', 'mnth_cos', 'weekday', 'workingday', 'holiday', 'weathersit', 'temp', 'hum', 'windspeed', 'yr']
target = ['cnt']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df[features], df[target], test_size=0.20, random_state=42
)
```

```python
# Hour, weekday and weather situation are categories, not quantities (hour 23 is not "more" than hour 1)
categorical_cols = ["hr", "weekday", "weathersit"]

categorical_encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)
train_categorical = categorical_encoder.fit_transform(X_train[categorical_cols])
test_categorical = categorical_encoder.transform(X_test[categorical_cols])
```

```python
from sklearn.preprocessing import StandardScaler

numeric_cols = ["mnth_sin", "mnth_cos", "temp", "hum", "windspeed", "yr", "workingday", "holiday"]

scaler = StandardScaler()
train_numeric = scaler.fit_transform(X_train[numeric_cols])
test_numeric = scaler.transform(X_test[numeric_cols])

X_train_encoded = np.column_stack([
    train_numeric,
    train_categorical
])

X_test_encoded = np.column_stack([
    test_numeric,
    test_categorical
])
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train_encoded, np.ravel(y_train))
```

```python
predictions = model.predict(X_test_encoded)

print(f"RMSE: {np.sqrt(mean_squared_error(y_test, predictions)):.2f}")
print(f"R²:   {r2_score(y_test, predictions):.3f}")
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    random_state=42,
    n_jobs=-1)
model.fit(X_train_encoded, np.ravel(y_train))
```

```python
predictions = model.predict(X_test_encoded)

print(f"RMSE: {np.sqrt(mean_squared_error(y_test, predictions)):.2f}")
print(f"R²:   {r2_score(y_test, predictions):.3f}")
```

```python
from sklearn.ensemble import HistGradientBoostingRegressor

model = HistGradientBoostingRegressor(
    learning_rate=0.05,
    max_iter=1000,
    max_leaf_nodes=31,
    l2_regularization=1.0,
    random_state=42
)

model.fit(X_train_encoded, np.ravel(y_train))
```

```python
predictions = model.predict(X_test_encoded)

print(f"RMSE: {np.sqrt(mean_squared_error(y_test, predictions)):.2f}")
print(f"R²:   {r2_score(y_test, predictions):.3f}")
```

```python
errors = np.abs(np.ravel(y_test) - predictions)

print(f"Within 25 rentals: {(errors <= 25).mean():.1%}")
print(f"Within 50 rentals: {(errors <= 50).mean():.1%}")
```
